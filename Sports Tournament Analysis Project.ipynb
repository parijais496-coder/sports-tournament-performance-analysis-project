```python
import pandas as pd

# Load dataset
df = pd.read_csv("WorldCupMatches[1].csv")

# Take team name from user
team = input("Enter team name: ")

# Filter matches
matches = df[(df["Home Team Name"] == team) |
              (df["Away Team Name"] == team)]

# Count matches
print("Total matches:", len(matches))

# Show team's matches
print(matches[["Home Team Name", "Home Team Goals",
               "Away Team Goals", "Away Team Name"]])

# Calculate total goals
goals = 0

for i in range(len(matches)):
    if matches.iloc[i]["Home Team Name"] == team:
        goals += matches.iloc[i]["Home Team Goals"]
    else:
        goals += matches.iloc[i]["Away Team Goals"]

print("Total goals scored:", goals)

# Top teams
print("\nTop 10 teams by matches:")
print(df["Home Team Name"].value_counts().head(10))

year = int(input("Enter World Cup year: "))
opponent = input("Enter opponent team name: ")

# Search by year
print("Matches in selected year:", len(matches[matches["Year"] == year]))

# Search by opponent
print("Matches against opponent:", len(matches[
    matches["Home Team Name"].str.contains(opponent, case=False, na=False) |
    matches["Away Team Name"].str.contains(opponent, case=False, na=False)
]))
```
